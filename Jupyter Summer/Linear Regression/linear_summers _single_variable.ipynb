```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn import linear_model
```

```python
df= pd.read_csv(r"D:\Dataset\canada_per_capita_income.csv",encoding='UTF-8')
```

```python
df = df.rename(columns={'per capita income (US$)':'Capital'})
```

```python
df.head()
```

```python
%matplotlib inline
plt.xlabel('year')
plt.ylabel('Capital(US $)')

plt.scatter(df.year,df.Capital,color='red',marker="+")
```

```python
model=linear_model.LinearRegression()
model.fit(df[['year']],df.Capital)
```

```python

model.predict([[1922]])
```

```python
m=model.coef_ #m
```

```python
b=model.intercept_ #b
```

```python
y=m*1922+b
print(y)
```
